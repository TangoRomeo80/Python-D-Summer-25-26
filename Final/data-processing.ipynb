```python
# print(df.isna())
print(df.isna().sum())
```

```python
# Calculate the count and percentage of missing values
missing_count = df.isna().sum()
missing_percentage = (df.isna().mean() * 100)
missing_report = pd.DataFrame({'Missing Count': missing_count, 'Missing Percentage': missing_percentage})
print(missing_report)
```

```python
missing_report = missing_report.sort_values(by='Missing Percentage', ascending=False)
print(missing_report)
```

```python
# What do we do with deck?
df['deck_known'] = df['deck'].notna().astype(int) # Create a new column 'deck_known' that indicates whether the deck value is known (1) or missing (0)
df.head()
```

```python
df = df.drop(columns=['deck'])
df.head()
```

```python
# Grouping age by passenger class
df.groupby(['pclass'])['age'].median()
```

```python
# Fill missing age values using corresponding group median
group_age_median = (
    df.groupby(['pclass'])['age'].transform('median')
)
df['age'] = df['age'].fillna(group_age_median)
# print(df['age'].isna().sum()) # Check if there are any missing values in the 'age' column after filling
```

```python
print(df.duplicated().sum()) # Check for duplicate rows in the dataframe
```

```python
df = df.drop_duplicates()
df.reset_index(drop=True, inplace=True)
```
