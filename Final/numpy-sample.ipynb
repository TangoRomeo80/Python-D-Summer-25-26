```python
import numpy as np
```

```python
age_list = [22, 38, 26, 35, 35, 27, 54, 2, 27, 14]
age_array = np.array(age_list)
print(type(age_array))  # Output: <class 'numpy.ndarray'>
print(age_array)  # Output: [22 38 26 35 35 27 54 2 27 14]
```

```python
# ndarray properties
print('Shape: ', age_array.shape)  # Output: (10,) 
print('Dimensions: ', age_array.ndim)  # Output: 1
print('Size: ', age_array.size)  # Output: 10
print('Data type: ', age_array.dtype)  # Output: int64 (or int32 depending on the system)
```
